```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

df = pd.read_csv("../viljami/propaganda-radar/data/data.csv")

tfidf = TfidfVectorizer()
X = tfidf.fit_transform(df["headline"])

X.shape
```

```python
words = tfidf.get_feature_names_out()

tfidf_scores = pd.DataFrame.sparse.from_spmatrix(
    X,
    columns=words
)

tfidf_scores
```
