```python
from pathlib import Path
import sys

import pandas as pd

sys.path.insert(0, str(Path.cwd()))

from data import load_data
from model import build_vectorizer

data = load_data("../data/processed/headlines.csv")
vectorizer = build_vectorizer()
X = vectorizer.fit_transform(data["headline"])

X.shape
```

```python
words = vectorizer.get_feature_names_out()
common_feature_indices = X.getnnz(axis=0).argsort()[-20:][::-1]
selected_words = words[common_feature_indices]
tfidf_sample = pd.DataFrame(
    X[:5, common_feature_indices].toarray(),
    columns=selected_words,
    index=data.index[:5],
)

tfidf_sample
```
